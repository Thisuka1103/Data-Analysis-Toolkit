#**9. Custom plotting**

```python
response = PLT.get_methods_info()
for method in response['response']:
    print(f"Method : {method['method']}")
    print(f"Signature : {method['signature']}")
    print(f"Description: {method['description']}")
    print("-" * 60)
```

```python
import json

data = json.dumps({"records": [
    {"category": "A", "value": 40},
    {"category": "B", "value": 60},
    {"category": "C", "value": 25}
]})

result = PLT.plot_bar_chart(x='category', y='value', title='Bar Chart', data=data)
PLT.display_image(result)
```

```python
result = PLT.plot_pie_chart(names='category', values='value', title='Pie Chart', data=data)
PLT.display_image(result)
```

```python
data_hist = json.dumps({"records": [
    {"value": v} for v in inspector.df.select_dtypes(include='number').iloc[:, 0].dropna().tolist()
]})

result = PLT.plot_histogram(x='value', title='Histogram', data=data_hist)
PLT.display_image(result)
```

```python
data_heat = json.dumps({"records": [
    {"row": "A", "col": "X", "val": 10},
    {"row": "A", "col": "Y", "val": 20},
    {"row": "B", "col": "X", "val": 30},
    {"row": "B", "col": "Y", "val": 40}
]})

result = PLT.plot_heat_map(values='val', index='row', columns='col', title='Heatmap', data=data_heat)
PLT.display_image(result)
```

```python
data_multi = json.dumps({"records": [
    {"group": "X", "metric1": 10, "metric2": 20},
    {"group": "Y", "metric1": 30, "metric2": 15},
    {"group": "Z", "metric1": 25, "metric2": 35}
]})

result = PLT.plot_multi_column_bar_graph(
    xLabel='group',
    value_vars=['metric1', 'metric2'],
    title='Multi Column Bar',
    data=data_multi
)
PLT.display_image(result)
```
